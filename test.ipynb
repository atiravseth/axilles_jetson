```python
from scipy.interpolate import PchipInterpolator
import numpy as np
import matplotlib.pyplot as plt
```

```python
PEAK_TORQUE = 120
TAU_PHASE_ARRAY = np.array([0.0, 0.20, 0.35, 0.52, 0.70, 1.0])
TAU_VAL_ARRAY = np.array([0.0, 0.0,  0.2,  1.0,  0.0,  0.0])
```

```python
phases  = TAU_PHASE_ARRAY
torques = TAU_VAL_ARRAY * PEAK_TORQUE
torque_profile = PchipInterpolator(phases, torques)
```

```python
# Plot this profile
x = np.linspace(0, 1, 100)
y = torque_profile(x)
plt.plot(x, y)
plt.xlabel('Phase')
plt.ylabel('Torque')
plt.title('Torque Profile')
plt.grid()
plt.show()
```
