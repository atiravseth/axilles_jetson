```python
from scipy.interpolate import PchipInterpolator
import numpy as np
import matplotlib.pyplot as plt
```

```python
PEAK_TORQUE = 0.7 # Nm

TAU_PHASE_ARRAY = np.array([0.0, 0.6, 0.7, 0.8, 0.9, 1.0])
TAU_VAL_ARRAY = np.array([0.0, 0.0, 0.2,  1.0,  0.2,  0.0])

phases  = TAU_PHASE_ARRAY
torques = TAU_VAL_ARRAY * PEAK_TORQUE 

torque_profile = PchipInterpolator(phases, torques)
```

```python
# pLOT THE TORQUE PROFILE
x = np.linspace(0, 1, 100)
y = torque_profile(x)
plt.plot(x, y)
plt.xlabel('Phase')
plt.ylabel('Torque (Nm)')
plt.title('Torque Profile')
plt.grid()
plt.show()
```
